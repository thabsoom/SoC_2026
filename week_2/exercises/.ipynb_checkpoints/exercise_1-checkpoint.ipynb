## Linear regression single variable

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn import linear_model
```

```python
df=pd.read_csv('./resources/canada_per_capita_income.csv')
df.head(3)
```

```python
%matplotlib inline
plt.xlabel('year')
plt.ylabel('per_capita_income(USD)')
plt.scatter(df['year'],df['per capita income (US$)'],color='red')
```

```python
lr=linear_model.LinearRegression()
lr.fit(df[['year']],df['per capita income (US$)'])
print("regression coefficient is" , lr.coef_)
print("regression intercept is" , lr.intercept_)
```

```python
predict_2020=lr.predict([[2020]])
print(predict_2020)
```

```python
#for plotting the regression line
new_df=df.drop('per capita income (US$)',axis='columns')
new=lr.predict(new_df)
new_df['per_capita']=new
```

```python
%matplotlib inline
plt.xlabel('year')
plt.ylabel('per_capita_income(USD)')
plt.scatter(df['year'],df['per capita income (US$)'],color='red')
plt.plot(new_df['year'],new_df['per_capita'],color='blue')
```
