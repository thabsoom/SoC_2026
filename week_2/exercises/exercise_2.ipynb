## Linear regression Multiple variables

```python
import pandas as pd
import numpy as np
from sklearn import linear_model
import math
```

```python
df=pd.read_csv('./resources/hiring.csv')
df
```

```python
#preprocessing
df.experience=df.experience.fillna('zero')
df
```

```python
#new_learing
from word2number import w2n
df['experience'] = df['experience'].apply(w2n.word_to_num)
```

```python
#preprocessing
median_score=math.floor(df['test_score(out of 10)'].median())
df['test_score(out of 10)']=df['test_score(out of 10)'].fillna(median_score)
df
```

```python
lr=linear_model.LinearRegression()
lr.fit(df[['experience','test_score(out of 10)','interview_score(out of 10)']],df['salary($)'])
print("Regression coefficients are",lr.coef_)
print("Regression intercept is",lr.intercept_)
```

```python
lr.predict([[2,9,6]])
```
