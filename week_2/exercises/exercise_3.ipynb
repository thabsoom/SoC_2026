## Gradient descent

```python
import pandas as pd
import numpy as np
import math
```

```python
df=pd.read_csv('./resources/test_scores.csv')
df
```

```python
def gradient_descent(x,y):
    m_curr=b_curr=0
    n=len(x)
    learning_rate=0.0001
    i=prev_cost=1
    while True:
        y_pred=m_curr*x+b_curr
        cost=(1/n)*sum(val**2 for val in (y-y_pred))
        if math.isclose(prev_cost,cost,rel_tol=1e-6):
            break
        md = -(2/n)*sum(x*(y-y_pred))
        bd = -(2/n)*sum(y-y_pred) 
        m_curr-=learning_rate*md
        b_curr-=learning_rate*bd
        print ("m {}, b {}, cost {} iteration {}".format(m_curr,b_curr,cost, i))
        i+=1
        prev_cost=cost
gradient_descent(df['math'],df['cs'])
```
